## Exploring the products csv file

```python
# import
url = "https://drive.google.com/file/d/1mbJgcLmn8OT-V3yJLgh3D5J50RV411sP/view?usp=drive_link" # products.csv
path = "https://drive.google.com/uc?export=download&id="+url.split("/")[-2]
product_df = pd.read_csv(path)
```

```python
#looking at the dataframe structure and features
product_df.info()
product_df.shape
product_df.describe(include='all')
#to have  a look at what the data looks like, tail could also be used
product_df.head()
# looking at how many unique
product_df['sku'].str[:3].nunique()
#looking for any NaN values
product_df.isna().sum()
```

```python
# looking at whether the promo price is always smaller than the price
(product_df['price'] > product_df['promo_price']).sum()
(product_df['price'] == product_df['promo_price']).sum()
```

```python
# looking at whether the non numeric columns can be converted
product_df['price'] = pd.to_numeric(product_df['promo_price'])
```

```python
# looking for NaN values
#product_df['type'].unique()
#product_df.index[product_df['type'].isna()]
product_df.loc[product_df['type'].isna(), 'type']
```

```python
# looking for any duplicate rows
product_df[product_df.duplicated()]
```

## Exploring the orders csv file

```python
# import
url = "https://drive.google.com/file/d/1J-XJS9NS6fXH-dA9KEurAJ-MqNhVJqEE/view?usp=drive_link" # orders.csv
path = "https://drive.google.com/uc?export=download&id="+url.split("/")[-2]
orders_df = pd.read_csv(path)
```

```python
# looking at the data
orders_df.head()
orders_df.info()
orders_df.isna().sum()
```

```python
# looking at the orders status and dominance
#orders_df.loc[orders_df['total_paid'].isna(),:]
orders_df.loc[~(orders_df['state'].isin(['Completed'])),:].shape[0]
```

```python
# looking at what typical order values are
#orders_df.loc[(orders_df['total_paid']== 0.000000) & (orders_df['state'] == 'Completed'),:].shape[0]
#orders_df.loc[(orders_df['total_paid']== 214747.530000),:]
orders_df.loc[orders_df['state'] == 'Completed',:].nlargest(10,'total_paid')
```

```python
# checking if the dates can be converted to datetime
orders_df['created_date'] = pd.to_datetime(orders_df['created_date'])
```

```python
# looking for duplicates
orders_df[orders_df.duplicated()]
```
